```python
def process_dataframe(df):
    df.index = df["Pockets"][["pdb", "pocket"]].apply(lambda x: "_".join(x), axis=1)
    df = df.drop(columns=["Pockets"], level=0)
    df.columns = map(lambda x: "_".join(x), df.columns.values)
    df.loc[:,'Label_label'] = df['Label_label'].astype("category")
    return df
```

```python
from sklearn.metrics import matthews_corrcoef, f1_score, confusion_matrix
def print_top1_results(modelfit, dataset):
    top1 = (
        modelfit.predict_proba(dataset)
        .assign(pdb=dataset.index.map(lambda x: x.split("_")[0]))
        .assign(
            pred=lambda df: df.groupby("pdb")[1].transform(
                lambda x: (x == x.max()).astype(int)
            )
        )
        ["pred"]
    )
    print( 
        "mcc", matthews_corrcoef(dataset["Label_label"], top1), "\n",
        "macro-f1", f1_score(dataset["Label_label"], top1, average="macro"), "\n",
        pd.DataFrame(confusion_matrix(dataset["Label_label"], top1))
    )
```

```python
print_top1_results(fit, testdataset)
```

```python
def print_topx_results(modelfit, dataset):
    top1 = (
        modelfit.predict_proba(dataset)
        .assign(pdb=dataset.index.map(lambda x: x.split("_")[0]))
        .assign(
            pred=lambda df: (
                # Start from a Series where each value/row (sample/pocket) is the total number of pos. labels on its PDB
                (
                    dataset
                    .assign(pdb=dataset.index.map(lambda x: x.split("_")[0]))
                    .groupby("pdb")["Label_label"].transform("sum")
                )
                # Then subtract this "total num. of pos. in a PDB" by the rank of each pocket in a PDB, sorted by the probability of 1
                .sub(df.groupby("pdb")[1].rank(method="first", ascending=False))
                # If the subtraction is positive or 0 it means that the pocket is in the topX and will be assigned 1
                >= 0).astype(int)
        )
        ["pred"]
    )
    print( 
        "mcc", matthews_corrcoef(dataset["Label_label"], top1), "\n",
        "macro-f1", f1_score(dataset["Label_label"], top1, average="macro"), "\n",
        pd.DataFrame(confusion_matrix(dataset["Label_label"], top1))
    )
```

```python
def print_topx_results(modelfit, dataset):
    y = (
        dataset.index.to_series()
        .str.split("_", n=1, expand=True)
        .rename(columns={0: "pdb", 1: "pocket"})
        .merge(pockets_apos[["pdb", "pocket", "site_in_pocket"]], on=["pdb", "pocket"], how="left")
        .set_index(dataset.index)["site_in_pocket"]
        .ge(0.35)
        .astype(int)
    )

    top1 = (
        modelfit.predict_proba(dataset)
        .assign(pdb=dataset.index.to_series().str.split("_", n=1).str[0])
        .assign(
            pred=lambda df: (
                # Start from a Series where each value/row (sample/pocket) is the total number of pos. labels on its PDB
                y.groupby(df["pdb"]).transform("sum")
                # Then subtract this "total num. of pos. in a PDB" by the rank of each pocket in a PDB, sorted by the probability of 1
                .sub(df.groupby("pdb")[1].rank(method="first", ascending=False))
                # If the subtraction is positive or 0 it means that the pocket is in the topX and will be assigned 1
                >= 0
            ).astype(int)
        )["pred"]
    )

    cm = confusion_matrix(y, top1)
    n0 = (y.groupby(dataset.index.to_series().str.split("_", n=1).str[0]).sum() == 0).sum()
    cm[1, 0] += n0  # add 1 FN for PDBs with no positive-labelled pockets

    print(
        "mcc", matthews_corrcoef(y, top1), "\n",
        "macro-f1", f1_score(y, top1, average="macro"), "\n",
        pd.DataFrame(cm)
    )
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde, pearsonr


def plot_holo_apo_feature(feature, testdataset, test_apos, pockets_apos, holosapos):
    label_col="Label_label"
    inv = {v: k for k, v in holosapos.items()}
    apo_labels = (
        pockets_apos.assign(
            _k=pockets_apos["pdb"].astype(str) + "_" + pockets_apos["pocket"].astype(str),
            _lab=(pockets_apos["site_in_pocket"] >= 0.35).astype(int),
        )
        .set_index("_k")["_lab"]
    )
    test_apos = test_apos.copy()
    test_apos[label_col] = apo_labels.reindex(test_apos.index).fillna(0).astype(int)

    apo_pos = test_apos.loc[test_apos[label_col] == 1].assign(pdb=lambda d: d.index.str.split("_").str[0])
    holo_pos = testdataset.loc[testdataset[label_col] == 1].assign(pdb=lambda d: d.index.str.split("_").str[0])

    pairs = []
    for apo_pdb, a in apo_pos.groupby("pdb", sort=False):
        holo_pdb = inv.get(apo_pdb)
        if not holo_pdb:
            continue
        h = holo_pos.loc[holo_pos["pdb"] == holo_pdb]
        if a.empty or h.empty:
            continue
        a = a.sort_values(feature)[feature].to_numpy()
        h = h.sort_values(feature)[feature].to_numpy()
        n = max(len(a), len(h))
        pairs.append(pd.DataFrame({"holo": np.resize(h, n), "apo": np.resize(a, n)}))

    df = pd.concat(pairs, ignore_index=True) if pairs else pd.DataFrame(columns=["holo", "apo"])

    fig, (ax0, ax1) = plt.subplots(1, 2, figsize=(11, 4.2), constrained_layout=True)
    if df.empty:
        ax0.text(0.5, 0.5, "No matched positive pockets.", ha="center", va="center")
        ax0.axis("off")
    else:
        ax0.scatter(df["holo"], df["apo"], s=18, alpha=0.8)
        lo = np.nanmin(df[["holo", "apo"]].to_numpy())
        hi = np.nanmax(df[["holo", "apo"]].to_numpy())
        ax0.plot([lo, hi], [lo, hi], linewidth=1)
        r, p = pearsonr(df["holo"], df["apo"])
        ax0.set_title(f"{feature}\nPearson r={r:.3f}, p={p:.2g}")
        ax0.set_xlabel("Holo")
        ax0.set_ylabel("Apo")

    rng = np.random.default_rng(0)
    x_h = df["holo"].dropna().to_numpy()
    x_a = df["apo"].dropna().to_numpy()
    y_h = rng.normal(0.0, 0.06, size=len(x_h))
    y_a = 1.0 + rng.normal(0.0, 0.06, size=len(x_a))
    ax1.scatter(x_h, y_h, s=14, alpha=0.7, label="holo")
    ax1.scatter(x_a, y_a, s=14, alpha=0.7, label="apo")

    xs = np.concatenate([x_h, x_a]) if (len(x_h) and len(x_a)) else (x_h if len(x_h) else x_a)
    if len(xs) > 1:
        grid = np.linspace(xs.min(), xs.max(), 300)
        for x, y0 in [(x_h, 0.0), (x_a, 1.0)]:
            if len(x) > 1:
                dens = gaussian_kde(x)(grid)
                dens = 0.35 * dens / dens.max() if dens.max() else dens
                ax1.plot(grid, y0 + dens, linewidth=1)
                ax1.fill_between(grid, y0, y0 + dens, alpha=0.18)

                m = x.mean()
                ax1.vlines(m, y0, y0 + 0.35, linestyles="--", linewidth=1)

    ax1.set_yticks([0, 1], ["holo", "apo"])
    ax1.set_xlabel(feature)
    ax1.set_title("Distribution")
    ax1.legend(frameon=False)

    return fig, (ax0, ax1)
```
