```python
import pandas as pd
from pathlib import Path
```

```python
news = tuple(pd.read_pickle("../../../training_data/7.Extra_set/pockets_features.pkl").keys())

news
```

```python
p = Path(".")#.resolve()
for pdb in news:
    if len(tuple(p.glob(f"{pdb}*"))) == 0:
        print(pdb)
```

```python
for pdb in news:
    for f in p.glob(f"{pdb}*"):
        if "_fail" not in f.name:
            print(pdb)
```

```python
for pdb in news:
    for f in p.glob(f"{pdb}*"):
        with open(f, "r") as fh:
            try:
                if "AFR results for:" in fh.read():
                    print(pdb)
            except:
                print(f)
                raise
```
